Bài 23.6: Dữ liệu mạng lưới giao thông đóng vai trò là nền tảng kỹ thuật giúp hệ thống AI đô thị thông minh có thể số hóa cấu trúc đường sá dưới dạng đồ thị để thực hiện các bài toán tối ưu hóa lộ trình và dự báo lưu lượng giao thông theo thời gian thực. Thông qua việc phân tích các thông số như mật độ nút giao và chiều dài các cạnh đường, AI có thể nhận diện các điểm nghẽn cổ chai để điều phối đèn tín hiệu thông minh, đồng thời hỗ trợ các nhà quản trị mô phỏng nhằm giảm thiểu ùn tắc và khí thải. Đây cũng là nguồn dữ liệu định vị thiết yếu giúp xe tự hành ra quyết định di chuyển an toàn và chính xác trong môi trường đô thị phức tạp.

```python
import osmnx as ox
import networkx as nx
import folium
center_point = (10.7713532, 106.670117)
G = ox.graph_from_point(center_point, dist=3000, network_type='drive')
m = folium.Map(location=center_point, zoom_start=13)
customers = [
    {'name': 'Customer 1', 'coords': (10.7561278, 106.6616627)},
    {'name': 'Customer 2', 'coords': (10.7713532, 106.670117)},
    {'name': 'Customer 3', 'coords': (10.7721351, 106.6995673)}
]
drivers = [
    {'name': 'Grab', 'coords': (10.7688907, 106.6704163), 'color': 'green'},
    {'name': 'Xanh', 'coords': (10.7572119, 106.6650421), 'color': 'blue'},
    {'name': 'Moovtek', 'coords': (10.7566964, 106.6597263), 'color': 'purple'}
]
assigned_drivers = []
print("Kêt quả tiến trình")
for cust in customers:
    cust_node = ox.distance.nearest_nodes(G, cust['coords'][1], cust['coords'][0])
    best_driver = None
    min_dist = float('inf')
    best_route = None
    for drv in drivers:
        if drv['name'] in assigned_drivers:
            continue
        drv_node = ox.distance.nearest_nodes(G, drv['coords'][1], drv['coords'][0])
        try:
            distance = nx.shortest_path_length(G, drv_node, cust_node, weight='length')
            if distance < min_dist:
                min_dist = distance
                best_driver = drv
                best_route = nx.shortest_path(G, drv_node, cust_node, weight='length')
        except:
            continue
    if best_driver and best_route:
        assigned_drivers.append(best_driver['name'])
        print(f" {best_driver['name']} ({best_driver['color']}) đến {cust['name']} sẽ có quãng đường là {round(min_dist, 1)}m")
        route_coords = [[G.nodes[node]['y'], G.nodes[node]['x']] for node in best_route]

        folium.PolyLine(route_coords,color=best_driver['color'], weight=6, opacity=0.7,tooltip=f"Lộ trình {best_driver['name']}").add_to(m)

for c in customers:
    folium.Marker(c['coords'], tooltip=c['name'], icon=folium.Icon(color='red', icon='user', prefix='fa')).add_to(m)
for d in drivers:
    folium.Marker(d['coords'], tooltip=d['name'], icon=folium.Icon(color=d['color'], icon='car', prefix='fa')).add_to(m)
m
```

```python
import osmnx as ox
import networkx as nx
import folium
place_point = (10.7713532, 106.670117)
G = ox.graph_from_point(place_point, dist=2000, network_type='drive')
for u, v, k, data in G.edges(data=True, keys=True):
    length = data.get('length', 0)
    if length < 80:
        data['risk_level'] = 'High'
        data['color'] = 'red'
    elif length < 150:
        data['risk_level'] = 'Medium'
        data['color'] = 'yellow'
    else:
        data['risk_level'] = 'Low'
        data['color'] = 'green'
start_coords = (10.765, 106.663)
end_coords = (10.775, 106.680)
orig_node = ox.distance.nearest_nodes(G, start_coords[1], start_coords[0])
dest_node = ox.distance.nearest_nodes(G, end_coords[1], end_coords[0])
route = nx.shortest_path(G, orig_node, dest_node, weight='length')
route_coords = [[G.nodes[node]['y'], G.nodes[node]['x']] for node in route]
m = folium.Map(location=place_point, zoom_start=13)
for u, v, k, data in G.edges(data=True, keys=True):
    points = [ [G.nodes[u]['y'], G.nodes[u]['x']], [G.nodes[v]['y'], G.nodes[v]['x']] ]
    folium.PolyLine(points, color=data['color'], weight=2, opacity=0.5).add_to(m)
folium.PolyLine(route_coords, color='blue', weight=6, opacity=0.9, tooltip="Tuyến đường đề xuất").add_to(m)
folium.Marker(start_coords, popup="Điểm đi", icon=folium.Icon(color='blue')).add_to(m)
folium.Marker(end_coords, popup="Điểm đến", icon=folium.Icon(color='red')).add_to(m)

m
```

```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMap
from sklearn.ensemble import RandomForestRegressor
from datetime import datetime
np.random.seed(42)
n_data = 500
lats = np.random.uniform(10.7688, 106.6704, n_data)
lons = np.random.uniform(10.7688, 106.6704, n_data)
hours = np.random.randint(0, 24, n_data)
demand = (1 / (np.abs(lats - 10.7688) + np.abs(lons - 10.6704) + 0.1)) * \
         (np.sin((hours - 8) * np.pi / 12) + 2) + np.random.normal(0, 1, n_data)
demand = np.maximum(0, demand)
df = pd.DataFrame({'lat': lats, 'lon': lons, 'hour': hours, 'demand': demand})
X = df[['lat', 'lon', 'hour']]
y = df['demand']
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)
predict_hour = 18
grid_lat, grid_lon = np.mgrid[10.768:10.769:20j, 10.669:10.670:20j]
grid_points = np.vstack([grid_lat.ravel(), grid_lon.ravel(),
                         np.full(grid_lat.size, predict_hour)]).T
predictions = model.predict(grid_points)
m = folium.Map(location=[10.7688, 106.6704], zoom_start=13)
heat_data = [[p[0], p[1], w] for p, w in zip(grid_points, predictions)]
HeatMap(heat_data, radius=15, blur=10, min_opacity=0.3).add_to(m)
title_html = f'''<h3 align="center" style="font-size:16px"><b>Dự báo nhu cầu gọi xe lúc {predict_hour}:00</b></h3>'''
m.get_root().html.add_child(folium.Element(title_html))
m
```
